```python
from pathlib import Path

from huggingface_hub import snapshot_download

REPO = "YoelRuso/MatriculasVC"
DATOS = Path.cwd() / "datos"   # el notebook debe ejecutarse desde la raíz del proyecto
PARTES = {
    "dataset": ["dataset_limpio/**"],
    "ocr": ["matriculas_recortadas/**"],
    "video": ["videos/**"],
    "originales": ["originales_roboflow/**"],
}

# None = todo (~540 MB); o una lista con partes, p. ej. ["dataset"] solo para entrenar
SOLO = ["dataset"]
```

```python
patrones = None if not SOLO else ["README.md"] + [p for s in SOLO for p in PARTES[s]]
snapshot_download(REPO, repo_type="dataset", local_dir=DATOS, allow_patterns=patrones)
print("Datos en", DATOS)
```
